```python
nlp = spacy.load('en_core_web_sm')

# Load list of common skills (can be extended)
SKILLS_DB = [
    "machine learning", "python", "java", "sql", "communication", "teamwork", "leadership", "teamlead","manager","scrum","sqa","word","access",
    "deep learning", "data analysis", "nlp", "project management", "excel", "pandas", "keras","unity","SFML","godot","c","c++", "tensorflow"
]

def extract_name(text):
    doc = nlp(text[:500])  # only top part
    for ent in doc.ents:
        if ent.label_ == "PERSON" and len(ent.text.split()) <= 3:
            return ent.text
    return None

# Extract text from PDF
def extract_pdf_text(pdf_path):
    return extract_text(pdf_path)

# Extract email using regex
def extract_email(text):
    match = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}", text)
    return match[0] if match else None

# Extract phone number using regex
def extract_phone(text):
    match = re.findall(r"\+?\d[\d -]{8,}\d", text)
    return match[0] if match else None

# E 
def extract_name(text):
    lines = [line.strip() for line in text.split("\n") if line.strip()]
    
    for line in lines[:10]:
        # Ignore lines with emails or numbers
        if "@" in line or any(char.isdigit() for char in line):
            continue
        
        words = line.split()
        if 2 <= len(words) <= 4:
            return line
    
    return None

# Extract skills from text
def extract_skills(text):
    text = text.lower()
    found = [skill for skill in SKILLS_DB if skill.lower() in text]
    return list(set(found))

# Extract education degrees
DEGREE_MAP = {
    "bachelor": "Bachelor's",
    "bs": "Bachelor's",
    "b.sc": "Bachelor's",
    "b.tech": "Bachelor's",
    "master": "Master's",
    "ms": "Master's",
    "m.sc": "Master's",
    "m.tech": "Master's",
    "phd": "PhD"
}

def extract_education(text):
    found = set()
    text = text.lower()

    for key, value in DEGREE_MAP.items():
        if re.search(rf"\b{key}\b", text):
            found.add(value)

    return list(found)

# Full resume parser
def parse_resume(pdf_path):
    text = extract_pdf_text(pdf_path)

    return {
        "Name": extract_name(text),
        "Email": extract_email(text),
        "Phone": extract_phone(text),
        "Skills": extract_skills(text),
        "Education": extract_education(text)
    }

# Example usage
resume_data = parse_resume(r"/Users/affifa/Downloads/Affifa Z Ahmad Resume.pdf")
print(resume_data)
```
